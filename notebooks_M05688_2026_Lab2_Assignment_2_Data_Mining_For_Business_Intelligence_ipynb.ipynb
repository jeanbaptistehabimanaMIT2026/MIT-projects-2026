```python

#  Task 4:

 # Look at a row where City == 'Rubavu'. What values do City_Kigali, City_Huye, City_Musanze, and City_Rubavu have for that row? Why?

# One-hot encode the city column
city_dummies =pd.get_dummies(df_clean['City'], prefix='City')

# Combine the new columns with the original dataframe
df_city = pd.concat([df_clean, city_dummies], axis=1)

df_city[df_city['City'] =='Rubavu'] [['City', 'City_Kigali', 'City_Huye', 'City_Musanze', 'City_Rubavu']].head(1)
```

```python
 For the row where City = 'Rubavu', the values are:

  . City_Kigali = False
  . City_Huye = False
  . City_Musanze = False
  . City_Rubavu  = True
Explanation: One-hot encoding creates a separate binary column for each city. Since the customer is from Rubavu,
             only the City_Rubavu column is True (equivalent to 1), while all other city columns are False (equivalent to 0).
This indicates that the customer belongs only to the Rubavu category.
```

```python
# Task 5: Write one line of code to manually normalize CustomerAge (without sklearn) and store it in Age _norm_manual.

df_clean['Age_norm_manual'] = (df_clean['CustomerAge'] - df_clean['CustomerAge'].min()) / (df_clean['CustomerAge'].max() - df_clean['CustomerAge'].min())
df_clean[['CustomerAge' , 'Age_norm_manual']].head()
```

```python
df_clean['Age_norm_manual'].describe()
```
